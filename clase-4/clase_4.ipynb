1. Misión: Análisis de la Urbanización
Problema del mundo real
Debido a la industrialización, existe una concentración urbana de la población en muchos países. Analizaremos información demográfica sobre 4 grandes ciudades para investigar la concentración poblacional.

Datos estadísticos (Unidad: miles de habitantes):
A: 9765

B: 3441

C: 2954

D: 1531

```python
# Misión Urbanización - Implementación procedural usando variables globales
def urban(city):
    global max_pop, min_pop, pop_sum
    n = 0
    for name, pop in city.items():
        if pop > max_pop:
            max_pop = pop
        if pop < min_pop:
            min_pop = pop
        pop_sum += pop
        n += 1
    print('población máxima:', max_pop)
    print('población mínima:', min_pop)
    print('diferencia entre población máxima y población mínima:', max_pop - min_pop)
    print('población media:', pop_sum / n)

# Variables globales iniciales
max_pop = 0
min_pop = 1000000
pop_sum = 0

city_pop = {'A': 9765, 'B': 3441, 'C': 2954, 'D': 1531}
urban(city_pop)
```

2. Regla de Ámbito (Scope Rule) en Python: LEGB
Python utiliza la regla LEGB para buscar la definición de una variable:

L (Local): Variables definidas dentro de la función activa.

E (Enclosing): Variables en la función contenedora (funciones anidadas).

G (Global): Variables declaradas a nivel superior del módulo/archivo.

B (Built-in): Nombres predefinidos integrados en el lenguaje (print, len, abs).


```python
# Demostración de la regla LEGB
x = 10  # Variable Global (G)
y = 11  # Variable Global (G)

def foo():
    x = 20  # Variable Enclosing (E) para bar / Local para foo
    def bar():
        a = 30  # Variable Local (L) para bar
        print("Valores (L, E, G):", a, x, y)
    bar()
    x = 40
    bar()

foo()
```

```python
# Demostración del peligro al sobreescribir funciones integradas (Built-in)

# Si descomentas estas líneas, romperás la función abs() nativa:
abs = 10
#print(abs(-5))  # Lanza TypeError: 'int' object is not callable

# Para solucionar si se sobreescribió:
del abs  # Elimina la variable global para restaurar el Built-in

print("Uso correcto de la función built-in abs(-10):", abs(-10))
```

```python
# Diferencia entre variables locales y la palabra clave global

def print_counter():
    counter = 200  # Variable local
    print('counter interno =', counter)

counter = 100  # Variable global
print_counter()
print('counter externo =', counter)

print("-" * 30)

def print_counter_global():
    global counter
    counter = 200  # Modifica la variable global externa
    print('counter interno (global) =', counter)

print_counter_global()
print('counter externo (modificado) =', counter)
```

3. Funciones de Primera Clase, Anidadas y la palabra clave nonlocal
En Python, las funciones son objetos de primera clase. Pueden asignarse a variables, guardarse en listas o devolverse en un return. Daremos paso a las funciones anidadas y al uso de nonlocal.

```python
# Funciones de Primera Clase 5
def plus(a, b):
    return a + b

def minus(a, b):
    return a - b

f_list = [plus, minus]
print('Suma mediante lista:', f_list[0](100, 200))
print('Resta mediante lista:', f_list[1](100, 200))

print("-" * 20)

# Funciones Anidadas
def decorate(style='italic'):
    def italic(s):
        return '<i>' + s + '</i>'
    def bold(s):
        return '<b>' + s + "1" + '</b>'

    if style == 'italic':
        return italic
    else:
        return bold

dec = decorate('italic')
print(dec('hello'))
dec2 = decorate('bold')
print(dec2('hello'))
```

```python
# Búsqueda de variables con 'nonlocal' - 6
def f():
    a = 777
    def g():
        a = 100
        def h():
            nonlocal a
            a = 333
        h()
        print("[Level 2] a =", a)
    g()
    print("[Level 1] a =", a)

f()
```

4. Closures (Cierres) y Closures con expresiones Lambda
Un Closure ocurre cuando una función anidada hace referencia a variables de su ámbito contenedor (free variables) y se devuelve dicha función. El entorno de variables se mantiene retenido en memoria incluso después de que la función externa termina su ejecución.

```python
# Closure básico con estados independientes -7
def makecounter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter

c1 = makecounter()
c2 = makecounter()

print('Contador c1:', c1())
print('Contador c1:', c1())
print('Contador c1:', c1())

print('Contador c1:', c1())

print('Contador c1:', c1())



print('Contador c2 (independiente):', c2())

print("-" * 30)

# Closure simplificado mediante Lambda
def closure_calc():
    a = 3
    b = 5
    return lambda x: a * x + b

c = closure_calc()
print('Evaluación Closure Lambda:', c(1), c(2), c(3), c(4), c(5))

print("-" * 30)

# Modificación de variable en Closure acumulativo
def calc_accum():
    a, b = 2, 3
    total = 0
    def mult_add(x):
        nonlocal total
        total = total + a * x + b
        return total
    return mult_add

c_acc = calc_accum()
print('Acumulado 1:', c_acc(1))
print('Acumulado 2:', c_acc(2))
print('Acumulado 3:', c_acc(3))
```

^Life: El símbolo ^ obliga a que el texto esté al inicio.

life$: El símbolo $ obliga a que esté al final.

[Ll]ife: Los corchetes permiten alternativas (acepta 'Life' o 'life').

\d{3}: \d representa un dígito numérico y {3} indica que se requieren 3 dígitos continuos.

match_obj.group(1): Extrae los subgrupos delimitados por paréntesis en la expresión regular.


```python
# Búsqueda y metacaracteres con el módulo 're' -9
import re

txt1 = "Life is too short, you need python."
txt2 = "The best moments of my life."

print('Búsqueda exacta "Life":', re.search('Life', txt1))
print('Búsqueda opciones "[Ll]ife":', re.search('[Ll]ife', txt2))
print('Metacarácter Inicio "^Life":', re.search('^Life', txt1))
print('Metacarácter Fin "life$":', re.search('life$', txt2))

print("-" * 30)

# Compilación de patrones y extracción de grupos
text = 'please call 010-2345-1234'
regex = re.compile(r'(\d{3})-(\d{4}-\d{4})')
match_obj = regex.search(text)

if match_obj:
    print('Coincidencia completa:', match_obj.group())
    print('Grupo 1 (prefijo):', match_obj.group(1))
    print('Grupo 2 (número):', match_obj.group(2))

# Extraer todas las ocurrencias
print('Findall:', re.findall(r'[Mm]y', 'My life my life my life in the sunshine'))
```

1. Misión: Sistema de Cuenta Bancaria (BankAccount)
Diseño de la clase BankAccount con encapsulación de datos, métodos de depósito, retiro y representación formal en texto.

```python
class Perro:
    # El método especial __init__ es el "constructor"
    # Se ejecuta automáticamente al crear un nuevo perro
    def __init__(self, nombre, edad):
        self.nombre = nombre  # Atributo: su nombre
        self.edad = edad      # Atributo: su edad

    # Un método: una acción que puede realizar el perro
    def ladrar(self):
        print(f"{self.nombre} dice: ¡Guau, guau!")

    def cumplir_anios(self):
        self.edad += 1
        print(f"¡Feliz cumpleaños {self.nombre}! Ahora tienes {self.edad} años.")
```

```python
# Creamos dos objetos (instancias) usando la clase Perro
mi_perro = Perro("Firulais", 3)
otro_perro = Perro("Luna", 5)

# Accedemos a sus características (atributos)
print(mi_perro.nombre)  # Imprime: Firulais
print(otro_perro.edad)   # Imprime: 5

# Les ordenamos realizar acciones (métodos)
mi_perro.ladrar()        # Imprime: Firulais dice: ¡Guau, guau!
otro_perro.ladrar()      # Imprime: Luna dice: ¡Guau, guau!

# Modificamos su estado
mi_perro.cumplir_anios() # Imprime: ¡Feliz cumpleaños Firulais! Ahora tienes 4 años.
```

```python
# Implementación de la Clase BankAccount
class BankAccount:
    """Ejemplo de clase en Python para modelar una cuenta bancaria"""
    def __init__(self, name, account_num, balance=0):
        self.name = name
        self.account_num = account_num
        self.balance = balance

    def get_name(self):
        return self.name

    def get_account_num(self):
        return self.account_num

    def get_balance(self):
        return self.balance

    def deposit(self, amount):
        self.balance += amount
        print('Se ha ingresado {} won. El saldo es de {} won.'.format(amount, self.balance))
        return self.balance

    def withdraw(self, amount):
        if self.balance - amount >= 0:
            self.balance -= amount
            return self.balance
        else:
            print('El saldo de la cuenta es de {} won. Es inferior a la cantidad de retirada solicitada {} won.'.format(self.balance, amount))

    def __str__(self):
        return 'El saldo de la cuenta {} de {} es {} won.'.format(self.account_num, self.name, self.balance)

# Prueba de ejecución de la Misión
account1 = BankAccount("David", "1234-0001")
print(account1)
account1.deposit(2000)
account1.deposit(500)

print(account1)
account1.withdraw(500)
account1.withdraw(5000)

```

3. Definición de Clases, self y Variables de Instancia vs Variables de Clase

```python
# Clase Cat y manipulación de instancias
class Cat:
    def __init__(self, name, color):
        self.name = name
        self.color = color

    def meow(self):
        print("Mi nombre es {}, color es {}, miau miau~~".format(self.name, self.color))

    def __str__(self):
        return "Cat(name={}, color={})".format(self.name, self.color)

# Creación de múltiples objetos/instancias
nabi = Cat('nabi', 'negro')
nero = Cat('nero', 'blanco')
mimi = Cat('mimi', 'marrón')

nabi.meow()
nero.meow()
mimi.meow()
```

```python
# Variables de Clase vs Variables de Instancia
class Circle:
    PI = 3.1415  # Variable de Clase (Compartida por todas las instancias)

    def __init__(self, name, radius):
        self.name = name      # Variable de Instancia
        self.radius = radius  # Variable de Instancia

    def area(self):
        return (self.radius ** 2) * Circle.PI

c1 = Circle('C1', 4)
c2 = Circle('C2', 6)

print("Área de c1:", c1.area())
print("Área de c2:", c2.area())
print("Atributos de c1 con __dict__:", c1.__dict__)
```

4. Encapsulación y Métodos Setters/Getters
Protección de los atributos internos contra asignaciones no válidas (p. ej., edades negativas).

```python
# Encapsulación mediante atributos privados (__nombre_atributo)
class EncapsulatedCat:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age  # Atributo privado

    def set_age(self, age):  # Setter con validación
        if age > 0:
            self.__age = age
        else:
            print("Edad inválida. Debe ser positiva.")

    def get_age(self):      # Getter
        return self.__age

    def __str__(self):
        return "Cat(nombre={}, edad={})".format(self.__name, self.__age)

cat_obj = EncapsulatedCat('nabi', -6)
print(cat_obj)

# Intentos de modificación
cat_obj.set_age(-5)  # Intento inválido
cat_obj.set_age(4)   # Intento válido
print(cat_obj)
```

5. Métodos Especiales (Sobrecarga de Operadores)
Permite personalizar la conducta de operadores del lenguaje (+, -, ==) en objetos propios.

```python
# Sobrecarga del operador '+' con __add__ en la clase Vector2D -15
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __str__(self):
        return "Vector2D({}, {})".format(self.x, self.y)

v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)
v3 = v1 + v2  # Llama internamente a v1.__add__(v2)
print("Resultado de la suma de vectores:", v3)
```

6. Herencia y Estructura Jerárquica
Mecanismo para reutilizar código definiendo subclases (hijas) basadas en superclases (padres).

```python
# Herencia: Clase Persona, Empleado y Gerente
class Person:
    def __init__(self, name):
        self.name = name

    def get_name(self):
        return self.name

class Employee(Person):
    def __init__(self, name, staff_id):
        super().__init__(name)  # Invocación del constructor de la clase padre
        self.staff_id = staff_id

    def info(self):
        return "Empleado: " + self.name + ", staffID: " + str(self.staff_id)

class Manager(Person):
    def __init__(self, name, position):
        super().__init__(name)
        self.position = position

    def info(self):
        return "Gerente: " + self.name + ", posición: " + str(self.position)

emp1 = Employee("David Doe", 1111)
emp2 = Employee("Paul Carter", 2222)
mgr = Manager("Anna Smith", "CFO")

print(emp1.info())
print(emp2.info())
print(mgr.info())
```
