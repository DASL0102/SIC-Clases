### 1. La Operación Básica: Intercambio (`swap`)
Los algoritmos de ordenamiento se basan en dos operaciones fundamentales: **comparar** e **intercambiar**.
* **Método tradicional (`swap1`)**: Usa una variable temporal `t`.
* **Método Pythonic (`swap2`)**: Usa empaquetado y desempaquetado de tuplas.

```python
# 1.1 Intercambio con variable temporal
def swap1(S, x, y):
    t = S[x]
    S[x] = S[y]
    S[y] = t

S = [10, 20]
print("Antes de swap1:", S)
swap1(S, 0, 1)
print("Después de swap1:", S)

print("-" * 30)

# 1.2 Intercambio al estilo Python (Tuplas)
def swap2(S, x, y):
    S[x], S[y] = S[y], S[x]

S = [10, 20]
print("Antes de swap2:", S)
swap2(S, 0, 1)
print("Después de swap2:", S)
```

---
### 2. Ordenamiento de Burbujas (Bubble Sort)
**Analogía de la Misión:** Tenemos 5 canicas de distinto peso `[50, 30, 40, 10, 20]` y un dispositivo que **solo puede comparar e intercambiar dos canicas adyacentes**.
* En cada pasada, comparamos pares adyacentes y enviamos el más pesado hacia la derecha.
* Al terminar la primera pasada, el elemento más grande (`50`) queda fijo al final.

```python
def bubblesort(S):
    n = len(S)
    for i in range(n):
        print(S)  # Visualizamos el estado en cada pasada i
        for j in range(n - 1):
            if S[j] > S[j + 1]:
                S[j], S[j + 1] = S[j + 1], S[j]

S = [50, 30, 40, 10, 20]
bubblesort(S)
print("Resultado final:", S)
```

---
### 3. Ordenamiento por Selección (Selection Sort)
**Analogía de la Misión:** Ahora tenemos una balanza que nos permite comparar canicas **que no son adyacentes**. Buscamos la canica más ligera de todas y la colocamos al principio (izquierda).

Veremos dos implementaciones:
1. **`selectionsort1(S)`**: Usando una lista adicional `R` (sin intercambio, extrayendo con `pop`).
2. **`selectionsort2(S)`**: Ordenamiento *in situ* (directamente sobre la lista `S` sin usar memoria extra).

```python
# 3.1 Ordenamiento por selección sin intercambio (con memoria adicional R)
def selectionsort1(S):
    R = []
    while len(S) > 0:
        print(R, S)
        smallest = S.index(min(S))
        R.append(S[smallest])
        S.pop(smallest)
    return R

S = [50, 30, 40, 10, 20]
R = selectionsort1(S)
print("Resultado final R:", R)
```

```python
# 3.2 Ordenamiento por selección IN SITU (sin memoria adicional)
def selectionsort2(S):
    n = len(S)
    for i in range(n - 1):
        print(S)
        smallest = i
        for j in range(i + 1, n):
            if S[j] < S[smallest]:
                smallest = j
        S[i], S[smallest] = S[smallest], S[i]

S = [50, 30, 40, 10, 20]
selectionsort2(S)
print("Resultado final S:", S)
```

---
### 4. Ordenamiento por Inserción (Insertion Sort)
**Analogía de la Misión:** Las canicas están dentro de un barril y **solo podemos sacar una a la vez**. A medida que sacamos una nueva canica, la comparamos secuencialmente con las que ya tenemos ordenadas afuera e insertamos la nueva en su posición correcta desplazando las mayores a la derecha.

Veremos dos implementaciones:
1. **`insertionsort1(S)`**: Usando una lista adicional `R`.
2. **`insertionsort2(S)`**: Ordenamiento *in situ* desplazando elementos dentro de `S`.

```python
# 4.1 Ordenamiento por inserción con memoria adicional R
def insertionsort1(S):
    n = len(S)
    R = []
    while len(S) > 0:
        print(R, S)
        x = S.pop(0)
        j = len(R) - 1
        while j >= 0 and R[j] > x:
            j -= 1
        R.insert(j + 1, x)
    return R

S = [50, 30, 40, 10, 20]
R = insertionsort1(S)
print("Resultado final R:", R)
```

```python
# 4.2 Ordenamiento por inserción IN SITU
def insertionsort2(S):
    n = len(S)
    for i in range(1, n):
        print(S)
        x = S[i]
        j = i - 1
        while j >= 0 and S[j] > x:
            S[j + 1] = S[j]
            j -= 1
        S[j + 1] = x

S = [50, 30, 40, 10, 20]
insertionsort2(S)
print("Resultado final S:", S)
```

---
### Un paso más: Análisis de Complejidad Temporal
* **Bubble Sort y Selection Sort:** Ejecutan un doble bucle `for` anidado. El número de comparaciones es:
  $$T(N) = (N - 1) + (N - 2) + \dots + 1 = \frac{N(N - 1)}{2} \implies O(N^2)$$
* **Insertion Sort:**
  * **Mejor caso (Lista ya ordenada):** Solo hace 1 comparación por pasada $\implies O(N)$.
  * **Peor caso (Lista en orden inverso):** Compara con todos los elementos anteriores en cada pasada $\implies O(N^2)$.

* **Q2.** ¿Cuántas operaciones de **comparación** (`S[j] > x`) se ejecutan al ordenar `S = [50, 30, 40, 10, 20]` con `insertionsort2(S)`?

```python
# Verificación en vivo de la Prueba Sorpresa Q1 y Q2
def bubblesort_quiz(S):
    swaps = 0
    n = len(S)
    for i in range(n):
        for j in range(n - 1):
            if S[j] > S[j + 1]:
                S[j], S[j + 1] = S[j + 1], S[j]
                swaps += 1
    return swaps

def insertionsort2_quiz(S):
    comparisons = 0
    n = len(S)
    for i in range(1, n):
        x = S[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if S[j] > x:
                S[j + 1] = S[j]
                j -= 1
            else:
                break
        S[j + 1] = x
    return comparisons

print("Q1 - Total de intercambios en Bubble Sort:", bubblesort_quiz([50, 30, 40, 10, 20]))
print("Q2 - Total de comparaciones en Insertion Sort:", insertionsort2_quiz([50, 30, 40, 10, 20]))
```

```python
def bubblesort_comparisons(S):
    # Copiamos la lista para no alterar la original
    S_copy = list(S)
    comparisons = 0
    n = len(S_copy)
    for i in range(n):
        for j in range(n - 1):
            comparisons += 1  # Cada vez que el if evalúa una relación, cuenta
            if S_copy[j] > S_copy[j + 1]:
                S_copy[j], S_copy[j + 1] = S_copy[j + 1], S_copy[j]
    return comparisons

def insertionsort_comparisons(S):
    S_copy = list(S)
    comparisons = 0
    n = len(S_copy)
    for i in range(1, n):
        x = S_copy[i]
        j = i - 1
        while j >= 0:
            comparisons += 1  # Cada iteración del while implica una comparación
            if S_copy[j] > x:
                S_copy[j + 1] = S_copy[j]
                j -= 1
            else:
                break
        S_copy[j + 1] = x
    return comparisons

# Lista de prueba
lista_prueba = [50, 30, 40, 10, 20]

print("Comparaciones en Bubble Sort:", bubblesort_comparisons(lista_prueba))
print("Comparaciones en Insertion Sort:", insertionsort_comparisons(lista_prueba))
```

### Detector de Anagramas
Un **anagrama** es una palabra formada reordenando las letras de otra palabra usando todas las letras originales exactamente una vez (ej. `"listen"` y `"silent"`).

**Reto:**
1. Crea una función `is_anagram` usando la función nativa `sorted()` de Python.
2. Crea una versión que ordene las letras usando nuestra función `selectionsort2()`.
3. Crea una versión que ordene las letras usando nuestra función `insertionsort2()`.

```python
# Solución Programación en Pareja - Unidad 27

# 1. Con sorted() nativo
def is_anagram_builtin(word1, word2):
    return sorted(word1) == sorted(word2)

# 2. Con selectionsort2()
def is_anagram_selection(word1, word2):
    list1, list2 = list(word1), list(word2)
    selectionsort2(list1)
    selectionsort2(list2)
    return list1 == list2

# 3. Con insertionsort2()
def is_anagram_insertion(word1, word2):
    list1, list2 = list(word1), list(word2)
    insertionsort2(list1)
    insertionsort2(list2)
    return list1 == list2

print("Pruebas con is_anagram_builtin:")
print(is_anagram_builtin("listen", "silent"))     # True
print(is_anagram_builtin("anagram", "nagaram"))   # True
print(is_anagram_builtin("listen", "silence"))    # False
print(is_anagram_builtin("anagram", "anagrams"))  # False
```

# Unidad 28: Ordenamiento por Fusión (Merge Sort)

### Objetivos de aprendizaje:
1. Comprender y aplicar el algoritmo **Merge Sort** bajo el paradigma **Divide y Vencerás** (usando recursividad).
2. Implementar la función de fusión (`merge`) que combina dos listas ordenadas en una sola lista ordenada.
3. Analizar el uso de memoria adicional ($2N$ vs $N$) entre `mergesort1` y `mergesort2`.

---
### 1. Misión y Concepto Clave
Si tenemos dos barriles de canicas que **ya están ordenados** (`[10, 20, 40, 80]` y `[30, 50, 60, 70]`), solo necesitamos comparar las canicas más ligeras de la parte superior de cada barril e ir pasándolas a un nuevo barril.

**Proceso de Merge Sort con `S = [27, 10, 12, 20, 25, 13, 15, 22]`:**
1. **Dividir:** Divide la lista por la mitad recursivamente hasta tener listas de 1 solo elemento (que por definición ya están ordenadas).
2. **Conquistar/Fusionar (`merge`):** Fusiona las sublistas de 1 en 2, de 2 en 4, y de 4 en 8 elementos ordenados.

```python
# 1. Implementación de Merge Sort con memoria adicional (Slicing L y R)
def merge1(S, L, R):
    k = 0
    while len(L) > 0 and len(R) > 0:
        if L[0] <= R[0]:
            S[k] = L.pop(0)
        else:
            S[k] = R.pop(0)
        k += 1
    while len(L) != 0:
        S[k] = L.pop(0)
        k += 1
    while len(R) != 0:
        S[k] = R.pop(0)
        k += 1

def mergesort1(S):
    n = len(S)
    if n > 1:
        print(S)
        mid = n // 2
        L, R = S[:mid], S[mid:]
        mergesort1(L)
        mergesort1(R)
        merge1(S, L, R)

S = [27, 10, 12, 20, 25, 13, 15, 22]
mergesort1(S)
print("Resultado final:", S)
```

---
### Un paso más: ¿Cuánta memoria adicional usa `mergesort1()`?
En cada llamada recursiva se crean nuevas listas `L` y `R` mediante *slicing*:
$$N + \frac{N}{2} + \frac{N}{2^2} + \dots + \frac{N}{2^k} \approx 2N$$
`mergesort1()` utiliza memoria adicional de tamaño **$2N$**.

### 2. Implementación de Merge Sort más eficiente (`mergesort2`)
En lugar de partir la lista creando copias `L` y `R`, pasamos los índices `low`, `mid` y `high` sobre la misma lista `S`, y solo usamos una lista auxiliar `R` durante el momento de la fusión. Esto reduce la memoria adicional a **$N$**.

```python
# 2. Implementación mejorada usando índices low, mid, high
def merge2(S, low, mid, high):
    R = []
    i, j = low, mid + 1
    while i <= mid and j <= high:
        if S[i] < S[j]:
            R.append(S[i]); i += 1
        else:
            R.append(S[j]); j += 1
    if i > mid:
        for k in range(j, high + 1):
            R.append(S[k]) # Nota: En las diapositivas aparece S[j], pero S[k] copia correctamente el resto del rango
    else:
        for k in range(i, mid + 1):
            R.append(S[k]) # Nota: En las diapositivas aparece S[i], pero S[k] copia correctamente el resto del rango
    for k in range(len(R)):
        S[low + k] = R[k]

def mergesort2(S, low, high):
    if low < high:
        print(S)
        mid = (low + high) // 2
        mergesort2(S, low, mid)
        mergesort2(S, mid + 1, high)
        merge2(S, low, mid, high)

S = [27, 10, 12, 20, 25, 13, 15, 22]
mergesort2(S, 0, len(S) - 1)
print("Resultado final:", S)
```

# Unidad 29: Ordenamiento Rápido (Quick Sort)

### Objetivos de aprendizaje:
1. Comprender **Quick Sort** como un algoritmo **Divide y Vencerás** que ordena **in situ** (sin requerir memoria adicional de tamaño $N$ como Merge Sort).
2. Implementar la función `partition` utilizando un **pivote** para enviar los elementos menores a la izquierda y los mayores a la derecha.
3. Analizar el peor caso de Quick Sort y cómo solucionarlo mediante un **pivote aleatorio** (`partition2`).

---
### 1. Implementación de Quick Sort con el Primer Elemento como Pivote (`quicksort1`)
Usamos la lista del material: `S = [15, 10, 12, 20, 25, 13, 22]`.
1. Elegimos `S[low]` (`15`) como `pivot`.
2. El índice `left` busca de izquierda a derecha el primer elemento **mayor** que el pivote (`20`).
3. El índice `right` busca de derecha a izquierda el primer elemento **menor** que el pivote (`13`).
4. Se intercambian `S[left]` y `S[right]`.
5. Cuando los índices se cruzan, intercambiamos el pivote con `S[pivotpoint]` para dejar el pivote en su posición definitiva.

```python
# 1.1 y 1.2 Quick Sort seleccionando el primer elemento como pivote
def partition1(S, low, high):
    pivot = S[low]
    left, right = low + 1, high
    while left < right:
        print(S)
        while left <= right and S[left] <= pivot:
            left += 1
        while left <= right and S[right] >= pivot:
            right -= 1
        if left < right:
            S[left], S[right] = S[right], S[left]
    pivotpoint = right
    S[low], S[pivotpoint] = S[pivotpoint], S[low]
    return pivotpoint

def quicksort1(S, low, high):
    if low < high:
        print(S)
        pivotpoint = partition1(S, low, high)
        quicksort1(S, low, pivotpoint - 1)
        quicksort1(S, pivotpoint + 1, high)

S = [15, 10, 12, 20, 25, 13, 22]
quicksort1(S, 0, len(S) - 1)
print("Resultado final:", S)
```

---
### 1.3 El Peor Caso de `quicksort1`: Listas ya ordenadas (ascendente o descendente)
Si la lista ya está ordenada (ej. `[10, 12, 13, 15, 20, 22, 25]` o en orden inverso `[25, 22, 20, 15, 13, 12, 10]`) y elegimos siempre el primer elemento como pivote, la partición no divide la lista en mitades equilibradas, sino que deja una parte vacía y otra con $N-1$ elementos, degradando el rendimiento a $O(N^2)$.

### 2. Solución: Quick Sort con Pivote Aleatorio (`quicksort2`)
Seleccionamos un índice aleatorio entre `low` y `high`, lo intercambiamos con `S[low]` y procedemos igual.

```python
from random import randint

def partition2(S, low, high):
    rand = randint(low, high)
    S[low], S[rand] = S[rand], S[low]
    pivot = S[low]
    left, right = low + 1, high
    while left < right:
        print(S)
        while left <= right and S[left] <= pivot:
            left += 1
        while left <= right and S[right] >= pivot:
            right -= 1
        if left < right:
            S[left], S[right] = S[right], S[left]
    pivotpoint = right
    S[low], S[pivotpoint] = S[pivotpoint], S[low]
    return pivotpoint

def quicksort2(S, low, high):
    if low < high:
        print(S)
        pivotpoint = partition2(S, low, high)
        quicksort2(S, low, pivotpoint - 1)
        quicksort2(S, pivotpoint + 1, high)

S = [25, 22, 20, 15, 13, 12, 10]
quicksort2(S, 0, len(S) - 1)
print("Resultado final:", S)
```
