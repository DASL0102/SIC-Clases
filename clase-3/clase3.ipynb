# 1. Unidad 17: Funciones y Modularidad
Una función es un bloque estructurado y reutilizable de código que realiza una tarea específica. Funciona como un componente de ensamble (análogo a piezas de Lego).

```python
# 1.1 Declaración básica y llamada
def print_star():
    print("************************")

print_star()
print_star()
```

```python
# 1.2 Parámetros vs. Argumentos
# 'n' es el parámetro (variable receptora)
def print_star_n(n):
    for _ in range(n):
        print("************************")

# '4' es el argumento (valor real pasado)
print_star_n(-2)
```

```python
# 1.3 Control de Errores de Tipo (TypeError por discrepancia de argumentos)
def print_sum(a, b):
    result = a + b
    print(f"La suma de {a} y {b} es {result}")

# Caso válido
print_sum(-10, 20)

# Descomentar la siguiente línea provocará un TypeError:
#print_sum(10)
```

```python
# 1.4 Parámetros Arbitrarios (*args)
# Empaqueta argumentos posicionales arbitrarios en una tupla
def sum_nums(*numbers):
    print("Tipo de dato recibido:", type(numbers), "| Contenido:", numbers)
    result = 0
    for n in numbers:
        result += n
    return result

print("Suma 1:", sum_nums(10, 20, 30))
print("Suma 2:", sum_nums(5, 15, 25, 35, 45))
```

```python
# 1.5 Parámetros por Defecto y Keyword Arguments
# Regla: Los parámetros por defecto deben definirse al final
def div(a=1, b=2):
    return a / b

print("div() con defaults:", div())        # a=1, b=2 -> 0.5
print("div(4) un argumento:", div(4))       # a=4, b=2 -> 2.0
print("div(6, 3) posicional:", div(6, 3))   # a=6, b=3 -> 2.0

# Keyword arguments: El orden de invocación no altera el mapeo
def get_root(a, b, c):
    discriminant = b**2 - 4 * a * c
    r1 = (-b + discriminant**0.5) / (2 * a)
    r2 = (-b - discriminant**0.5) / (2 * a)
    return r1, r2

print("Invocación posicional:", get_root(1, 2, -8))
print("Invocación por palabra clave:", get_root(c=-8, a=1, b=2))
print("Invocación combinada:", get_root(1, 2, c=-8))
```

```python
# 1.6 Ámbito (Scope): Variables Globales vs. Locales
a = 10
b = 20

def modify_scope():
    global a, b  # Permite modificar directamente las variables globales
    a = 100
    b = 200
    result = a + b
    print(f"[Dentro de la función] a: {a}, b: {b}, suma: {result}")

print(f"[Antes de ejecutar] a: {a}, b: {b}")
modify_scope()
print(f"[Fuera de la función] a: {a}, b: {b}")
```

# 2. Unidad 18: Recursión y Memorización
La recursividad consiste en resolver un problema dividiéndolo en instancias más pequeñas de sí mismo mediante llamadas a la propia función.
Todo algoritmo recursivo exige:
1. **Caso Base:** Condición de parada que detiene las llamadas.
2. **Paso Recursivo:** Avance hacia el caso base.

```python
# 2.1 Factorial Recursivo: n! = n * (n-1)!
def factorial(n):
    if n <= 1:  # Caso Base
        return 1
    else:       # Paso Recursivo
        return n * factorial(n - 1)

print("5! =", factorial(10))
```

```python
# 2.2 Problema Clásico: La Torre de Hanoi
def hanoi(n, desde, via, hasta):
    if n == 1:
        print(f"Mover disco 1 de {desde} -> {hasta}")
    else:
        hanoi(n - 1, desde, hasta, via)
        print(f"Mover disco {n} de {desde} -> {hasta}")
        hanoi(n - 1, via, desde, hasta)

print("Resolución de Torre de Hanoi (3 discos):")
hanoi(3, 'A', 'B', 'C')
```

```python
# 2.3 Fibonacci y el Costo del Doble Cómputo
import time

# Implementación recursiva ingenua (Exponencial O(2^n))
def fib_naive(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return fib_naive(n - 1) + fib_naive(n - 2)

# Implementación con Memorización (Lineal O(n))
memo = {0: 0, 1: 1}
def fib_memo(n):
    if n not in memo:
        memo[n] = fib_memo(n - 1) + fib_memo(n - 2)
    return memo[n]

start = time.time()
val_memo = fib_memo(100)
print(f"Fibonacci(35) con Memorización: {val_memo} (Tiempo: {time.time() - start:.6f}s)")

# start = time.time()
# val_naive = fib_naive(40)
# print(f"Fibonacci(35) Naive: {val_naive} (Tiempo: {time.time() - start:.6f}s)")
```

# 3. Unidad 19: Lambdas, Funciones de Orden Superior e Iteradores

```python
# 3.1 Expresiones Lambda
# Sintaxis: lambda parámetros: expresión
add = lambda x, y: x + y
print("Suma con lambda:", add(100, 200))
print("Ejecución inmediata:", (lambda x, y: x + y)(100, 200))
```

```python
# 3.2 filter(), map() y reduce()
from functools import reduce

# filter: filtrar elementos que cumplan una condición booleana
ages = [34, 39, 20, 18, 13, 54]
adults = list(filter(lambda x: x >= 19, ages))
print("Mayores de edad (>=19):", adults)

#map: transformar cada elemento de una colección
numbers = [1, 2, 3, 4, 5, 6, 7]
squares = list(map(lambda x: x**2, numbers))
print("Cuadrados:", squares)

# reduce: acumular valores aplicando una función de dos argumentos
suma_acumulada = reduce(lambda acc, val: acc + val, [1, 2, 3, 4, 5])
print("Suma acumulada con reduce:", suma_acumulada)
```
