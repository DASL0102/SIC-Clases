## 📌 Unidad 14: Métodos de Diccionarios (Parte 1) y Manipulación de Cadenas
Los diccionarios permiten almacenar asociaciones clave-valor. A continuación revisamos los métodos principales para actualizar, consultar y eliminar elementos.

```python
# ==========================================================
# PARTE A: Conteo Tradicional (con IF) vs Conteo con setdefault
# ==========================================================
str1 = ['a', 'b', 'c', 'b', 'a', 'b', 'c']

# 1. Enfoque tradicional: requiere validar si la clave existe
dic_tradicional = {}
for ch in str1:
    if ch not in dic_tradicional:
        dic_tradicional[ch] = 0
    dic_tradicional[ch] += 1

print("--- COMPARACIÓN 1: Inicialización y Conteo ---")
print("Resultado con IF clásico:   ", dic_tradicional)
```

```python
# 2. Enfoque con setdefault: inicializa en 0 solo si la clave no existe
dic_setdefault = {}
for ch in str1:
    dic_setdefault.setdefault(ch, 0)
    dic_setdefault[ch] += 1

print("--- COMPARACIÓN 1: Inicialización y Conteo ---")
print("Resultado con setdefault(): ", dic_setdefault)
print("¿Producen el mismo resultado?:", dic_tradicional == dic_setdefault)

```

```python

# ==========================================================
# PARTE B: Choque directo setdefault() vs update()
# ==========================================================
print("\n--- COMPARACIÓN 2: ¿Qué pasa cuando la clave YA EXISTE? ---")

# Caso con setdefault(): NO modifica claves existentes
d1 = {'a': 10, 'b': 20}
print("Diccionario inicial d1:        ", d1)
retorno_set = d1.setdefault('a', 999) # 'a' ya existe con valor 10
print("d1.setdefault('a', 999) retorna:", retorno_set)
print("Estado final de d1 (no cambió):", d1)

# Caso con update(): SÍ sobreescribe claves existentes
d2 = {'a': 10, 'b': 20}
print("\nDiccionario inicial d2:        ", d2)
d2.update(a=999) # 'a' ya existe, aquí sí se fuerza el cambio
print("d2.update(a=999) aplicado:")
print("Estado final de d2 (sí cambió):", d2)

```

```python
d2.update(a=90, e=50)
print(d2)
```

```python

# ==========================================================
# PARTE C: Sintaxis de update() (Strings vs Claves Numéricas)
# ==========================================================
print("\n--- COMPARACIÓN 3: Formas de usar update() ---")
dic_datos = {'nombre': 'David', 1: 'nivel_1'}

# Para strings podemos usar argumentos por nombre clave=valor
dic_datos.update(nombre='Carlos', rol='Admin')

# Para claves que NO son strings (números, tuplas), usamos lista de pares o un diccionario
dic_datos.update([[1, 'nivel_top'], [2, 'nivel_medio']])
dic_datos.update([[3,'nivel_bajo']])
print("Resultado final tras ambos update():", dic_datos)
```

### Manipulación de Texto para Procesamiento de Datos

Métodos esenciales para limpieza, sanitización y formateo de cadenas (`str`):

| Método | Sintaxis | Descripción | Ejemplo de Entrada | Resultado |
| :--- | :--- | :--- | :--- | :--- |
| **`strip()`** | `s.strip([chars])` | Elimina espacios en blanco (o caracteres indicados) al inicio y al final. | `"  contacto@mail.com \n".strip()` | `'contacto@mail.com'` |
| **`split()`** | `s.split([sep])` | Divide la cadena en una lista según el delimitador especificado (por defecto, espacios). | `"Juan,Pérez,30".split(",")` | `['Juan', 'Pérez', '30']` |
| **`join()`** | `sep.join(iterable)` | Une los elementos de una lista/iterable en una sola cadena usando `sep` como conector. | `"-".join(["2026", "10", "15"])` | `'2026-10-15'` |
| **`replace()`** | `s.replace(old, new)` | Reemplaza todas las apariciones de una subcadena por otra. | `"$1,250.00".replace("$", "")` | `'1,250.00'` |
| **`startswith()`** | `s.startswith(prefix)` | Retorna `True` si la cadena inicia con el prefijo indicado; de lo contrario, `False`. | `"+58-412-1234567".startswith("+58")` | `True` |
| **`endswith()`** | `s.endswith(suffix)` | Retorna `True` si la cadena finaliza con el sufijo indicado; de lo contrario, `False`. | `"reporte_octubre.csv".endswith(".csv")` | `True` |

---

#### 💡 Casos de uso comunes:
* **`strip()`**: Limpiar saltos de línea (`\n`), tabulaciones (`\t`) y espacios accidentales de archivos planos o web scraping.
* **`split()`**: Parsear columnas compuestas (por ejemplo: nombres completos, direcciones, logs del sistema).
* **`join()`**: Reensamblar textos tras filtrar *stop words* o construir rutas y fechas con formato estándar.
* **`replace()`**: Eliminar caracteres monetarios (`$`, `€`), símbolos de porcentaje o comas antes de transformar a `float`/`int`.
* **`startswith()` / `endswith()`**: Filtrado de rutas de archivos por extensión (`.csv`, `.parquet`) o validación de códigos de área.

```python
# ==========================================
# 3. Métodos de Cadenas
# ==========================================
texto_sucio = "   Actions \n\t speak louder than words   "

# Limpieza combinando split y join
palabras = texto_sucio.split()
print(palabras)
texto_limpio = " ".join(palabras)
print("Texto depurado:", texto_limpio)

```

```python
# strip / lstrip / rstrip con caracteres
cadena_num = "########ejemplo_datos#####"
print("Strip personalizado:", cadena_num.strip('#'))
```

```python

# startswith / endswith / replace
archivo = "2026_reporte_ventas.csv"
print("¿Empieza con 2026?", archivo.startswith('2026'))
print("¿Es CSV?", archivo.endswith('.csv'))

precios = "100,200,300,400"
print("Replace (remover comas):", precios.replace(",", " "))
```

## 📌 Unidad 15: Métodos de Diccionarios (Parte 2), Copias y Módulos Estándar
Veremos cómo estructurar diccionarios anidados, evitar errores con `defaultdict`, diferenciar copias superficiales de copias profundas, y utilizar módulos clave de la biblioteca estándar (`datetime`, `time`, `math`).

```python
# ==========================================
# 4. dict.fromkeys() y collections.defaultdict
# ==========================================
from collections import defaultdict

claves = ['id', 'nombre', 'rol']
nuevo_dic = dict.fromkeys(claves, "Desconocido")
print("Diccionario desde fromkeys:", nuevo_dic)


```

```python
# defaultdict evita KeyError
contador = defaultdict(int) #el argumento int es la función fábrica (default_factory) que define qué valor asignar automáticamente cuando intentas acceder o modificar una clave que todavía no existe en el diccionario.
for char in 'abcbaabca':
    contador[char] += 1
print("Contador automático:", dict(contador))


```

```python
# defaultdict con listas
agrupador = defaultdict(list)
agrupador['frutas'].append('manzana')
agrupador['frutas'].append('pera')
agrupador['frutas'].append(1)
print("Agrupador de listas:", dict(agrupador))
```

### Diccionarios Anidados y Copia de Objetos
- Asignación (`y = x`): Ambas variables apuntan a la misma dirección de memoria.
- `copy()`: Copia superficial (shallow copy). Modificar subdiccionarios afectará al original.
- `copy.deepcopy()`: Copia recursiva e independiente en memoria.

```python
# ==========================================
# 5. Diccionarios dobles y Deepcopy
# ==========================================
import copy

datos_sistema = {
    'nodo_1': {'ip': '192.168.1.10', 'estado': 'activo'},
    'nodo_2': {'ip': '192.168.1.20', 'estado': 'inactivo'}
}

# Copia superficial vs Profunda
copia_superficial = datos_sistema.copy()
copia_profunda = copy.deepcopy(datos_sistema)

# Modificamos el subdiccionario en la copia profunda
copia_profunda['nodo_1']['estado'] = 'mantenimiento'

print("Original nodo_1:", datos_sistema['nodo_1']['estado']) # Sigue siendo 'activo'
print("Copia profunda nodo_1:", copia_profunda['nodo_1']['estado'])
```

```python
copia_superficial['nodo_2']['estado'] = 'mantenimiento'
print("Original nodo_1:", datos_sistema['nodo_2']['estado']) # Sigue siendo 'activo'
print("Copia superficial nodo_1:", copia_superficial['nodo_2']['estado'])
```

## 📌 Unidad 16: Conjuntos (`set`) y la Función `zip()`
Un `set` es una colección no ordenada de elementos únicos (sin duplicados). Soporta operaciones algebraicas de conjuntos como unión, intersección, diferencia y diferencia simétrica.

```python
# ==========================================
# 7. Operaciones con Conjuntos (set)
# ==========================================
s1 = {1, 2, 3, 4, 5, 6}
s2 = {4, 5, 6, 7, 8, 9}

print("Unión (s1 | s2):", s1 | s2)
print("Intersección (s1 & s2):", s1 & s2)
print("Diferencia (s1 - s2):", s1 - s2)
print("Diferencia simétrica (s1 ^ s2):", s1 ^ s2)

# Set comprehensions (filtrar elementos)
fruta = 'pineapple'
letras_filtradas = {char for char in fruta if char not in 'apl'}
print(f"Letras de '{fruta}' sin 'a', 'p', 'l':", letras_filtradas)
```

### Agrupación y Emparejamiento con `zip()`
`zip(*iterables)` toma elementos iterables y genera tuplas correspondientes de igual índice hasta que el iterable más corto se agote.

```python
# ==========================================
# 8. Función zip() y Desempaquetado
# ==========================================
nombres = ["David", "John", "Jane"]
edades = [25, 30, 22]
cursos = ["Python", "Bases de Datos", "Redes", "Cálculo"] # Elemento extra ignorado

emparejados = list(zip(nombres, edades, cursos))
print(emparejados)
print("Lista combinada con zip:")
for n, e, c in emparejados:
    print(f"- Alumno: {n} | Edad: {e} | Curso: {c}")

# Desempaquetado (Unzipping)
nombres_recup, edades_recup, cursos_recup = zip(*emparejados)
print("\nDesempaquetado de nombres:", nombres_recup)
print(edades_recup)
```
