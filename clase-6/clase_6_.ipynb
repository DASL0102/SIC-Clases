## 1. Unidad 25: Búsqueda Binaria Básica y Evaluación de Rendimiento
El algoritmo de **Búsqueda Binaria** opera sobre listas previamente **ordenadas**. En cada paso, compara el elemento en la posición central (`mid`) con el valor buscado (`x`), reduciendo el espacio de búsqueda a la mitad.

```python
def bin_search(nums, x):
    """
    Realiza una búsqueda binaria en la lista ordenada 'nums' para encontrar 'x'.
    Imprime la traza de ejecución (low, high, mid) para analizar la complejidad temporal.
    Devuelve el índice de 'x' si existe, o -1 si no se encuentra.
    """
    low, high = 0, len(nums) - 1
    operaciones = 0
    print(f"\n[Iniciando Búsqueda Binaria] Buscando x = {x} en la lista de tamaño N = {len(nums)}")
    print(f"{'Paso':<6} | {'low':<5} | {'high':<5} | {'mid':<5} | {'nums[mid]':<10} | {'Decisión'}")
    print("-" * 60)

    while low <= high:
        operaciones += 1
        mid = (low + high) // 2
        valor_mid = nums[mid]

        if valor_mid == x:
            print(f"{operaciones:<6} | {low:<5} | {high:<5} | {mid:<5} | {valor_mid:<10} | ¡Encontrado!")
            print(f" Total de comparaciones realizadas: {operaciones}")
            return mid
        elif valor_mid > x:
            print(f"{operaciones:<6} | {low:<5} | {high:<5} | {mid:<5} | {valor_mid:<10} | {valor_mid} > {x} -> Mover high a {mid - 1}")
            high = mid - 1
        else:
            print(f"{operaciones:<6} | {low:<5} | {high:<5} | {mid:<5} | {valor_mid:<10} | {valor_mid} < {x} -> Mover low a {mid + 1}")
            low = mid + 1

    print(f" Búsqueda finalizada. Elemento no encontrado tras {operaciones} comparaciones.")
    return -1

# Lista ordenada de prueba (Página 20 del PDF)
S = [11, 17, 26, 28, 37, 45, 53, 59]

# Caso A: Elemento en la mitad (Mejor caso: 1 operación)
pos_mejor = bin_search(S, 28)

# Caso B: Elemento intermedio (3 operaciones)
pos_53 = bin_search(S, 53)

# Caso C: Elemento inexistente (Peor caso: 4 operaciones)
pos_peor = bin_search(S, 77)
```

## 2. Unidad 25: Problema del Mundo Real - La Caída de Huevos (Infinite Eggs)
Dado un edificio de `height` pisos y un piso `breaking` a partir del cual los huevos se rompen, determinamos el piso más alto en el que un huevo **NO** se rompe usando búsqueda binaria ($O(\log_2 N)$).

```python
def do_experiment(floor, breaking):
    """
    Devuelve True si el huevo SE ROMPE en el piso dado (floor >= breaking).
    Devuelve False si el huevo permanece intacto (piso seguro).
    """
    return floor >= breaking

def find_highest_safe_floor2(height, breaking):
    """
    Encuentra el piso más alto donde el huevo no se rompe mediante Búsqueda Binaria.
    """
    low, high = 1, height
    paso = 1
    print(f"\n[Iniciando Experimento de Huevos] Edificio de {height} pisos. Piso de ruptura real: {breaking}")
    print(f"{'Paso':<6} | {'low':<5} | {'high':<5} | {'mid':<5} | {'¿Se Rompe?':<12} | {'Nuevo Rango Espacio de Búsqueda'}")
    print("-" * 70)

    while low < high:
        mid = (low + high) // 2
        rompe = do_experiment(mid, breaking)

        if rompe:
            print(f"{paso:<6} | {low:<5} | {high:<5} | {mid:<5} | {'SÍ (Rompe)':<12} | high = {mid}")
            high = mid
        else:
            print(f"{paso:<6} | {low:<5} | {high:<5} | {mid:<5} | {'NO (Seguro)':<12} | low = {mid + 1}")
            low = mid + 1
        paso += 1

    piso_seguro = low - 1
    print(f"\nResultado Final: El huevo estará a salvo hasta el piso número {piso_seguro}.")
    return piso_seguro

# Simulación 1: Edificio de 100 pisos con punto de rotura en el piso 50 (Página 45 del PDF)
find_highest_safe_floor2(100, 50)

# Simulación 2: Edificio de 100 pisos con punto de rotura en el piso 51 (Página 46 del PDF)
find_highest_safe_floor2(100, 51)
```

## 3. Unidad 25: Resolución Interactiva de Quizzes y Pair Programming Q1
Evaluación paso a paso de los cuestionarios del PDF y el problema de inserción ordenada.

```python
def simular_juego_adivinar(maximum, number):
    """
    Simula el juego 'Adivinar Número' mediante Búsqueda Binaria (Quizzes 1 y 2, Páginas 48-49).
    """
    count = 0
    low, high = 1, maximum
    print(f"\n[Simulador Juego] Rango [1, {maximum}], Número secreto: {number}")

    while low < high:
        mid = (low + high) // 2
        count += 1
        print(f" Intento #{count}: low={low}, high={high}, mid={mid}")
        if mid == number:
            print(f" -> ¡Tu número es: {number}!")
            break
        elif mid > number:
            high = mid - 1
        else:
            low = mid + 1

    print(f" Total de veces que se realizó la búsqueda (count): {count}")
    return count

print("--- QUIZ #1 (maximum=100, number=51) ---")
simular_juego_adivinar(100, 51)

print("\n--- QUIZ #2 (maximum=100, number=25) ---")
simular_juego_adivinar(100, 25)

# # ==========================================
# # Unidad 25
# # ==========================================
# def search_insert_position(nums, x):
#     """
#     Dada una lista ordenada 'nums' y un elemento 'x', retorna el índice
#     donde debe insertarse 'x' manteniendo el orden ascendente.
#     """
#     low, high = 0, len(nums) - 1
#     while low <= high:
#         mid = (low + high) // 2
#         if nums[mid] == x:
#             return mid
#         elif nums[mid] > x:
#             high = mid - 1
#         else:
#             low = mid + 1
#     return low

# print("\n--- PROGRAMACIÓN EN PAREJA U25 - Q1 ---")
# nums_p53 = [10, 20, 40, 50, 60, 80]
# x_insert = 30
# pos_ins = search_insert_position(nums_p53, x_insert)
# print(f"En la lista {nums_p53}, el elemento {x_insert} debe ser ingresado en la posición {pos_ins}")
# nums_p53.insert(pos_ins, x_insert)
# print(f"Lista actualizada: {nums_p53}")
```

## 4. Unidad 26: Implementación de Tabla Hash Completa (Manejo de Colisiones por Encadenamiento)
Estructura orientada a objetos para almacenar pares clave-valor en buckets, utilizando aritmética modular para hashing.

```python
class HashTable:
    """
    Implementación de Tabla Hash según especificaciones del SIC (Página 73 del PDF).
    """
    def __init__(self, size):
        self.size = size
        self.table = {}
        for i in range(size):
            self.table[i] = []

    def hash(self, key):
        """Función hash básica mediante módulo sobre el tamaño de la tabla."""
        return key % self.size

    def put(self, key, value):
        """Almacena el valor en el bucket correspondiente si no existe previamente."""
        hash_key = self.hash(key)
        bucket = self.table[hash_key]
        if value not in bucket:
            bucket.append(value)
            print(f" [put] Clave ASCII: {key:<5} | HashKey (Index): {hash_key} | Guardado: '{value}'")
        else:
            print(f" [put] El valor '{value}' ya se encuentra en el Bucket {hash_key}.")

    def get(self, key):
        """Extrae el bucket asociado a la clave hash."""
        hash_key = self.hash(key)
        return self.table[hash_key]

    def mostrar_tabla(self):
        """Imprime el estado visual de los buckets."""
        print(f"\n--- Estado de la Tabla Hash (Tamaño: {self.size}) ---")
        for k in sorted(self.table.keys()):
            print(f"Bucket {k}: {self.table[k]}")

# Ejemplo del PDF: Estantería de Libros (Página 71-80)
ht_8 = HashTable(8)
libros = [
    "The Little Prince",
    "The Old Man and the Sea",
    "The Little Mermaid",
    "Beauty and the Beast",
    "The Last Leaf"
]

print("--- Almacenando Libros en la Tabla Hash (8 Buckets) ---")
for book in libros:
    clave_ascii = sum(map(ord, book))
    ht_8.put(clave_ascii, book)

ht_8.mostrar_tabla()

# Búsqueda por clave (Página 81)
buscar_titulo = "The Last Leaf"
key_leaf = sum(map(ord, buscar_titulo))
bucket_obtenido = ht_8.get(key_leaf)
print(f"\nBuscando '{buscar_titulo}' (Clave ASCII: {key_leaf}) -> Bucket {ht_8.hash(key_leaf)}: {bucket_obtenido}")
```

## 6. Unidad 26: Conversor de Números Romanos a Arábigos mediante Tablas Hash
Aplicación práctica de diccionarios en Python como tablas hash para traducir números romanos.

```python
# Tabla Hash de mapeo directo (Página 83 del PDF)
table_roman = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}

def roman_to_int(s):
    """
    Convierte una cadena de número romano a entero arábigo (Página 84-86 del PDF).
    """
    result = 0
    pasos_detalles = []

    for i in range(len(s) - 1):
        val_actual = table_roman[s[i]]
        val_siguiente = table_roman[s[i + 1]]

        if val_actual < val_siguiente:
            result -= val_actual
            pasos_detalles.append(f"-{val_actual} ({s[i]}) < {val_siguiente} ({s[i+1]})")
        else:
            result += val_actual
            pasos_detalles.append(f"+{val_actual} ({s[i]})")

    val_ultimo = table_roman[s[-1]]
    result += val_ultimo
    pasos_detalles.append(f"+{val_ultimo} ({s[-1]})")

    print(f" [Conversión] {s:<10} -> Desglose: {' '.join(pasos_detalles)} = {result}")
    return result

print("--- PRUEBAS DE CONVERSIÓN ROMANO A ARÁBIGO ---")
roman_to_int("MDCLXVI")   # Esperado: 1666 (Página 61)
roman_to_int("MCDXLIV")   # Esperado: 1444 (Página 62)
roman_to_int("MCMXCIX")   # Esperado: 1999 (Página 63)
```

## 7. Unidad 26: Programación en Parejas - Conversor Inverso (Arábigo a Romano)
Implementación de la solución solicitada en la **Página 93 del PDF**.

```python
def int_to_roman(num):
    """
    Convierte un entero arábigo a número romano usando la tabla hash invertida (Página 93).
    """
    table_inv = {
        1000: 'M', 900: 'CM', 500: 'D', 400: 'CD',
        100: 'C', 90: 'XC', 50: 'L', 40: 'XL',
        10: 'X', 9: 'IX', 5: 'V', 4: 'IV', 1: 'I'
    }

    num_original = num
    roman_str = ""

    for value in table_inv.keys():
        while num >= value:
            roman_str += table_inv[value]
            num -= value

    return roman_str

print("--- PROGRAMACIÓN EN PAREJA U26: ARÁBIGO A ROMANO ---")
ejemplos_pdf = [1999, 369, 80, 29, 155, 14, 492, 348, 301, 469, 499]
for n in ejemplos_pdf:
    print(f"Arábigo: {n:<5} -> Romano: {int_to_roman(n)}")
```
