- **push**: Agrega un elemento a la parte superior (top).
- **pop**: Remueve y retorna el elemento superior.
- **peek**: Consulta el elemento superior sin removerlo.
"""

```python
# ==========================================
# CELDA 3: [CÓDIGO] Clase Stack en Python (POO)
# ==========================================
class Stack:
    def __init__(self):
        self.stack = []

    def is_empty(self):
        return len(self.stack) == 0

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        if self.is_empty():
            return None
        return self.stack.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self.stack[-1]

    def size(self):
        return len(self.stack)

    def __str__(self):
        return f"Stack({self.stack})"

# Prueba del objeto Stack
s = Stack()
s.push("Banana")
s.push("Apple")
s.push("Tomato")
print("Pila cargada:", s)
print("Pop efectuado:", s.pop())
print("Pila resultante:", s)
```

```python
# ==========================================
# CELDA 4: [CÓDIGO] Aplicación 1: Verificador de Paréntesis Balanceados
# ==========================================
def check_parentheses(expr):
    opening = "[{("
    closing = "]})"  # Mismo orden: [ -> ], { -> }, ( -> )
    stack = Stack()

    for char in expr:
        if char in opening:
            stack.push(char)
        elif char in closing:
            if stack.is_empty():
                return False
            top_item = stack.pop()
            if opening.index(top_item) != closing.index(char):
                return False

    return stack.is_empty()

# Banco de Pruebas
ejemplos = [
    "[]{}()",
    "({[]})",
    "((())",
    "([()})",
    ")("
]

for exp in ejemplos:
    resultado = "Válido" if check_parentheses(exp) else "Inválido"
    print(f"Cadena: {exp:10} -> Resultado: {resultado}")
```

```python
# ==========================================
# CELDA 5: [CÓDIGO] Aplicación 2: Inversión de Texto usando Pila
# ==========================================
def reverse_string(text):
    s = Stack()
    for char in text:
        s.push(char)

    inverted = ""
    while not s.is_empty():
        inverted += s.pop()

    return inverted

texto_original = "Estructuras de Datos"
print(f"Original: {texto_original}")
print(f"Invertido: {reverse_string(texto_original)}")
```


## Módulo 2: La Cola (Queue)
Una **Cola** es una estructura de datos que opera bajo el principio **FIFO** (*First-In, First-Out* / Primero en Entrar, Primero en Salir).
- **enqueue**: Añade un elemento al final de la cola.
- **dequeue**: Remueve y retorna el primer elemento de la cola.
"""

```python
# ==========================================
# CELDA 7: [CÓDIGO] Clase Queue en Python
# ==========================================
class Queue:
    def __init__(self):
        self.queue = []

    def is_empty(self):
        return len(self.queue) == 0

    def enqueue(self, item):
        self.queue.append(item)

    def dequeue(self):
        if self.is_empty():
            return None
        return self.queue.pop(0)

    def size(self):
        return len(self.queue)

    def __str__(self):
        return f"Queue({self.queue})"

# Prueba de uso
q = Queue()
q.enqueue("Persona 1")
q.enqueue("Persona 2")
q.enqueue("Persona 3")
print("Cola inicial:", q)
print("Atendiendo:", q.dequeue())
print("Cola restante:", q)
print(q.size())
```

```python
# ==========================================
# CELDA 8: [CÓDIGO] Aplicación 1: El Problema de Josefo
# ==========================================
def josephus_sequence(n, k):
    sequence = []
    queue = Queue()

    # Insertar N personas numeradas del 1 a N
    for i in range(1, n + 1):
        queue.enqueue(i)

    step = 1
    while queue.size() > 1:
        person = queue.dequeue()
        if step % k == 0:
            sequence.append(person)
        else:
            queue.enqueue(person)
        step += 1

    # La última persona en la cola sobrevive
    survivor = queue.dequeue()
    sequence.append(survivor)

    return sequence, survivor

n, k = 7, 3
secuencia, sobreviviente = josephus_sequence(n, k)
print(f"Secuencia de eliminación para N={n}, K={k}: {secuencia}")
print(f"El sobreviviente final es la persona #{sobreviviente}")
```

```python
# ==========================================
# CELDA 9: [CÓDIGO] Aplicación 2: Simulación de Fila de Impresión
# ==========================================
import time

class PrinterQueue:
    def __init__(self):
        self.queue = Queue()

    def add_job(self, doc_name):
        print(f"[+] Documento '{doc_name}' añadido a la cola de impresión.")
        self.queue.enqueue(doc_name)

    def process_jobs(self):
        print("\n--- Iniciando impresión ---")
        while not self.queue.is_empty():
            current_doc = self.queue.dequeue()
            print(f"Imprimiendo: {current_doc}...")
        print("--- Todos los trabajos han sido procesados ---")

printer = PrinterQueue()
printer.add_job("Informe_Final.pdf")
printer.add_job("Grafico_Estadistico.png")
printer.add_job("Carta_Presentacion.docx")
printer.process_jobs()
```


## Módulo 3: Búsqueda Secuencial (Linear Search)
La **Búsqueda Secuencial** inspecciona cada elemento de un arreglo uno a uno, desde el índice `0` hasta `n-1`, hasta encontrar el elemento deseado o recorrer toda la lista.

- Complejidad Temporal Peor Caso: $O(n)$
- Complejidad Temporal Mejor Caso: $O(1)$
"""

```python
# ==========================================
# CELDA 11: [CÓDIGO] Búsqueda Secuencial Básica y Mayor Clave
# ==========================================
def sequential_search(nums, target):
    for i in range(len(nums)):
        if nums[i] == target:
            return i  # Retorna el índice donde se encontró
    return -1  # Retorna -1 si no existe

def find_largest_key(nums):
    if not nums:
        return -1
    largest_index = 0
    for i in range(1, len(nums)):
        if nums[i] > nums[largest_index]:
            largest_index = i
    return largest_index

# Pruebas
datos = [11, 37, 45, 26, 59, 28, 17, 53]
print("Lista de datos:", datos)
print("Índice de la clave 45:", sequential_search(datos, 45))
print("Índice de una clave inexistente (88):", sequential_search(datos, 88))
print("Índice del valor máximo en la lista:", find_largest_key(datos))
print("Valor máximo guardado en ese índice:", datos[find_largest_key(datos)])
```

```python
# ==========================================
# CELDA 12: [CÓDIGO] Aplicación: Experimento de la Caída de Huevos
# ==========================================
import random

def find_highest_safe_floor(total_floors):
    # Genera aleatoriamente el piso crítico donde se romperá el huevo
    critical_floor = random.randint(1, total_floors)

    def is_egg_broken(floor_tested):
        return floor_tested >= critical_floor

    # Búsqueda secuencial piso por piso
    for current_floor in range(1, total_floors + 1):
        if is_egg_broken(current_floor):
            return current_floor - 1, critical_floor

    return total_floors, critical_floor

floors = 100
safe_floor, break_floor = find_highest_safe_floor(floors)
print(f"Piso crítico simulado (se rompe en): {break_floor}")
print(f"El piso más alto donde el huevo estará a salvo es: {safe_floor}")
```
