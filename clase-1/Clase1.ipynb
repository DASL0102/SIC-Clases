#### 1. ¿Por qué necesitamos listas? (Misión Big Data)
Si tenemos las puntuaciones de varios exámenes o alturas de personas, declarar una variable para cada elemento genera código repetitivo, difícil de mantener y propenso a errores:

```python
# Variables individuales (ineficiente)
score0 = 87
score1 = 84
score2 = 95
score3 = 67
score4 = 88
score5 = 94
score6 = 63
print("Variables individuales:", score0, score3)

```

```python
# Solución: Estructura de Lista
# Una lista contiene múltiples elementos en una sola variable, separados por comas entre corchetes []
score_list = [87, 84, 95, 67, 88, 94, 63]
print("Lista:", score_list)
print("Acceso a elementos individuales:", score_list[0], score_list[3])
```

#### 2. Características de las Listas
- Pueden contener diferentes tipos de datos (enteros, números reales, cadenas) en una misma lista.
- Se pueden generar secuencias automáticas usando la función `range()`.

```python
# Lista de cadenas
fruits = ['banana', 'apple', 'orange', 'kiwi']
print("Frutas:", fruits)

```

```python
# Lista mixta (diferentes tipos de datos)
mixed_list = [100, 200, 'apple', 400]
print("Lista mixta:", mixed_list)

```

```python
# Crear listas numéricas continuas con range()
list4 = list(range(1, 10))
print("Lista con range(1, 10):", list4)
```

#### 3. Indexación e Indexación Negativa
- La indexación positiva va de `0` hasta `len(lista) - 1`.
- Python admite **índices negativos**, donde `-1` representa el último elemento, `-2` el penúltimo, etc.
- Acceder a un índice igual o mayor a `len(lista)` produce un `IndexError: list index out of range`.

```python
n_list = [11, 22, 33, 44, 55, 66]

print("Lista:", n_list)
print("Longitud len(n_list):", len(n_list))
tam = len(n_list)

```

```python
print(n_list[tam-1])
```

```python
# Índices positivos
print("Primer elemento n_list[0]:", n_list[0])
print("Segundo elemento n_list[1]:", n_list[1])

```

```python
# Índices negativos
print("Último elemento n_list[-1]:", n_list[-1])
print("Penúltimo elemento n_list[-2]:", n_list[-2])

```

```python
# Funciones incorporadas: len, max, min, sum, any
scores = [200, 700, 500, 300, 400]
print("Longitud:", len(scores))
print("Máximo:", max(scores))
print("Mínimo:", min(scores))
print("Suma:", sum(scores))
print("¿Hay algún elemento distinto de cero? (any):", any(scores))
```

#### 4. Añadir y Eliminar Elementos en Listas
- **`append(x)`**: Inserta un elemento al final.
- **`extend(iterable)`**: Concatena otra lista o elementos al final.
- **`insert(índice, x)`**: Inserta un valor en la posición indicada.
- **`del lista[i]`**: Instrucción clave de Python para eliminar por índice.
- **`pop()`**: Elimina y devuelve el último elemento (o en la posición dada).
- **`remove(x)`**: Elimina la primera aparición del valor especificado (genera `ValueError` si no existe).

```python
avengers = ["Iron Man", "Thor", "Hulk", "Capitán América"]
print("Original:", avengers)

```

```python
# append: une a Black Widow al final
avengers.append("Black Widow")
print("Después de append('Black Widow'):", avengers)

```

```python
# insert: coloca a Hawkeye en la posición con índice 1
avengers.insert(1, "Hawkeye")
print("Después de insert en índice 1:", avengers)

```

```python
# pop: extrae al último miembro de la lista
ultimo = avengers.pop()
print(f"Elemento extraído con pop(): {ultimo}")
print("Lista tras pop():", avengers)

```

```python
# del: elimina el elemento en el índice 0
del avengers[0]
print("Después de del avengers[0]:", avengers)

```

```python
# Operador 'in' para evitar errores antes de remove
if "Hulk" in avengers:
    avengers.remove("Hulk")
    print("Hulk removido con éxito:", avengers)
```

#### 5. Slicing (Rebanado / Troceado)
Sintaxis: `lista[inicio:fin:paso]`
- Corta desde el índice `inicio` hasta `fin - 1`.
- Cortar genera una copia parcial sin modificar la lista original.

```python
a_list = [10, 20, 30, 40, 50, 60, 70, 80]

print("Original:       ", a_list)
print("a_list[1:5]:    ", a_list[1:5])      # Desde índice 1 hasta 4
print("a_list[:5]:     ", a_list[:5])       # Desde el inicio hasta índice 4
print("a_list[1:]:     ", a_list[1:])       # Desde índice 1 hasta el final
print("a_list[:]:      ", a_list[:])        # Toda la lista
print("a_list[-7:-2]:  ", a_list[-7:-2])   # Con índices negativos
print("a_list[::2]:    ", a_list[::2])      # Saltando de 2 en 2
print("a_list[::-1]:   ", a_list[::-1])     # Lista en orden inverso
```

#### 6. Tuplas: Estructura Inmutable
- Definidas con paréntesis `( )` o valores separados por coma.
- Una tupla de un solo elemento requiere coma obligatoria: `(1,)`.
- **Inmutabilidad:** No se pueden reasignar ni eliminar elementos individualmente una vez creada (genera `TypeError`). Permite un acceso más rápido y seguro.

```python
tup_uno = (100,)
print("Tipo de (100,):", type(tup_uno))

colors = ("red", "green", "blue")
print("Tupla de colores:", colors)

```

#### 7. Diccionarios (Unidad 11)
- Estructura de pares clave-valor (`clave : valor`).
- Se utiliza la clave en lugar de un índice numérico para acceder o modificar los valores.

```python
# Creación del diccionario
persona = {
    'Nombre': 'David Doe',
    'Edad': 26,
    'Peso': 82
}

print("Diccionario persona:", persona)
print("Acceso por clave ['Nombre']:", persona['Nombre'])

# Modificación e inserción
persona['Edad'] = 27               # Modificar
persona['Trabajo'] = 'Empresa A'   # Insertar nuevo par clave-valor
print("Diccionario actualizado:", persona)

# Comprobar pertenencia de claves
print("¿Está 'Edad' en persona?:", 'Edad' in persona)
print("¿Está 'Ciudad' en persona?:", 'Ciudad' in persona)
```

#### 8. Misión Integradora Unidad 10: Base de Datos Personal y Slicing
Tenemos 4 personas integradas en una sola lista plana con el siguiente formato de 5 datos continuos por persona: `[nombre, edad, sexo, altura, peso]`.
Debemos calcular la edad media extrayendo las edades mediante rebanado con paso (`step = 5`).

```python
# Misión 10.3
person1 = ['David Doe', 20, 1, 180.0, 100.0]
person2 = ['John Smith', 25, 1, 170.0, 70.0]
person3 = ['Jane Carter', 22, 0, 169.0, 60.0]
person4 = ['Peter Kelly', 40, 1, 150.0, 50.0]

# Combinar listas con operador +
person_list = person1 + person2 + person3 + person4
print(person_list)

# Cada persona tiene 5 atributos
n_persons = int(len(person_list) / 5)
age_sum = 0.0

# person_list[1::5] empieza en el índice 1 (edad) y salta de 5 en 5
print("Edades extraídas con slicing [1::5]:", person_list[1::5])

for age in person_list[1::5]:
    age_sum += age

average_age = float(age_sum) / n_persons
print("La edad media es de " + str(average_age) + " años.")
```

#### 9. Misión Integradora Unidad 11: Menú de Cafetería y Formateo
Mostrar el menú de la cafetería con precios formateados y simular la selección de una bebida validando si la clave existe.

```python
# Misión 11.2 del documento
menu = {
    'Americano': 3000,
    'Ice Americano': 3500,
    'Cappuccino': 4000,
    'Cafe Latte': 4500,
    'Espresso': 3600
}

# Impresión formateada del menú
print("--- MENÚ CAFETERÍA ---")
for key in menu:
    print("{:16s} Precio: {:,} KRW".format(key, menu[key]))

# Consulta de bebida
eleccion = 'Cafe Latte'  # Puede cambiarse por input() si se interactúa en vivo

if eleccion in menu.keys():
    print("\n{} cuesta {:,} KRW. Por favor, realice el pago.".format(eleccion, menu[eleccion]))
else:
    print("\nLo sentimos. {} no está en el menú.".format(eleccion))
```
