## Unidad 30: Enfoque Greedy (Voraz)

El **enfoque Greedy** es una técnica de diseño de algoritmos que llega a una solución seleccionando la **"mejor" elección del momento** (óptimo local) en cada paso, esperando llegar a la mejor solución global.

### Los 3 pasos del diseño Greedy:
1. **Procedimiento de selección (Selection procedure):** Elegir el mejor elemento actual según nuestro criterio y añadirlo al conjunto.
2. **Comprobación de la viabilidad (Feasibility check):** Verificar si el elemento seleccionado no viola las condiciones del problema (ej. que no se pase del monto a devolver o que las reuniones no se solapen).
3. **Comprobación de la solución (Solution check):** Determinar si ya completamos la respuesta al problema.

## 1. Problema del Cambio de Moneda (Coin Change Problem)

En Corea del Sur existen cuatro denominaciones de monedas: **500, 100, 50 y 10 won**.
Queremos devolver un cambio (por ejemplo, **870 won**) entregando la **menor cantidad de monedas posible**, asumiendo que tenemos monedas infinitas de cada tipo.

**Pseudocódigo Greedy:**
```text
while (queden monedas y el problema no esté resuelto) {
    tomar la moneda restante más grande;                 // 1. Selección
    if (agregar la moneda no supera el cambio debido)    // 2. Viabilidad
        agregar la moneda al cambio;
    if (el valor total iguala al cambio debido)          // 3. Solución
        el problema está resuelto;
}

````python
### 🔹 CELDA 3 [Código] — Implementación de `coin_change`

def coin_change(coins, amount):
    changes = []
    largest = 0
    while amount > 0:
        if amount < coins[largest]:
            largest += 1
        else:
            changes.append(coins[largest])
            amount -= coins[largest]
    return changes

# Ejecución de prueba con 870 won
coins = [500, 100, 50, 10]
amount = 870  # Puedes cambiarlo por int(input("Input the amount: "))
changes = coin_change(coins, amount)
print(f"Monedas seleccionadas: {changes}")
print(f"Cantidad total de monedas: {len(changes)}")
````

¿Cuándo falla un algoritmo Greedy?

````python
# Supongamos que el banco central emite una nueva moneda de 400 won.
# ¿Qué devuelve nuestro algoritmo Greedy para un cambio de 800 won?

coins_quiz = [500, 400, 100, 50, 10]
amount_quiz = 800
changes_quiz = coin_change(coins_quiz, amount_quiz)
print("Resultado Greedy para 800 won:", changes_quiz, "-> Total monedas:", len(changes_quiz))
````

## 2. Problema de Selección de Actividades (Activity Selection Problem)

Una empresa tiene **una sola sala de conferencias** y recibe $N = 7$ solicitudes de equipos para usarla. Cada solicitud tiene una hora de **Inicio (`start`)** y una de **Fin (`finish`)**.
Queremos asignar la sala para celebrar el **máximo número de reuniones posibles** sin que se solapen.

| ID Reunión (`i`) | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: | :---: |
| **Inicio (`start`)** | 1 | 3 | 2 | 0 | 5 | 8 | 5 |
| **Fin (`finish`)** | 2 | 4 | 5 | 6 | 6 | 9 | 9 |

### ¿Cuál es el Criterio Greedy correcto?
Para dejar la sala libre lo antes posible para las demás reuniones, debemos **ordenar las reuniones de menor a mayor según su hora de FINALIZACIÓN (`finish`)** y elegir siempre la siguiente reunión que termine más temprano y no choque con la anterior.

Implementación de activity_selection1

````python
def activity_selection1(start, finish):
    result = []
    i = 0
    result.append(i)
    for j in range(1, len(start)):
        if finish[i] <= start[j]:
            result.append(j)
            i = j
    return result

# Datos de las 7 reuniones (ya ordenadas por tiempo de finalización 'finish')
start = [1, 3, 2, 0, 5, 8, 5]
finish = [2, 4, 5, 6, 6, 9, 9]

meetings = activity_selection1(start, finish)
maximum = len(meetings)
print(f"IDs de reuniones seleccionadas: {meetings}")
print(f"Número máximo de reuniones: {maximum}")
````

Cambio con Monedas Limitadas

````python
# Problema: Ahora NO tenemos monedas infinitas. Recibimos exactamente las monedas
# que hay en nuestra billetera: [500, 50, 50, 100, 50, 10, 10] y debemos dar 710 won.

def coin_change2(coins, amount):
    changes = []
    # Recorremos cada moneda física disponible en la billetera (ya ordenada de mayor a menor)
    for coin in coins:
        if amount >= coin:
            changes.append(coin)
            amount -= coin
        if amount == 0:
            break
    return changes

# Datos de prueba del ejercicio (Simulando el input: "500 50 50 100 50 10 10")
raw_input_coins = "500 50 50 100 50 10 10"
coins = list(map(int, raw_input_coins.split()))
coins.sort(reverse=True)  # Pista: Ordenar de forma descendente
print("Monedas en billetera (ordenadas):", coins)

amount = 710
changes = coin_change2(coins, amount)
print("Cambio devuelto:", changes, "-> Total monedas usadas:", len(changes))
````

# Unidad 31: Divide y Conquista (Divide and Conquer)

La estrategia **Divide y Conquista** resuelve un problema grande dividiéndolo recursivamente en subproblemas más pequeños hasta que sean tan simples que se resuelvan directamente.

### Los 3 pasos de Divide y Conquista:
1. **Dividir (Divide):** Dividir la instancia de entrada original en dos o más instancias más pequeñas.
2. **Conquistar (Conquer):** Resolver cada subproblema recursivamente. Si ya es lo suficientemente pequeño (caso base), resolverlo de inmediato.
3. **Combinar (Combine):** Si es necesario, integrar las soluciones pequeñas para obtener la respuesta al problema original.

### Comparación de algoritmos clásicos que usan Divide y Conquista:
* **Búsqueda Binaria (Unidad 25):** Divide la lista a la mitad, pero **solo conquista 1 de las 2 mitades** (descarta la otra). No necesita fase de combinación.
* **Merge Sort (Unidad 28):** Divide en 2 mitades **iguales**, conquista ambas y **necesita una fase de combinación (`merge`)** para fusionar las listas ordenadas.
* **Quick Sort (Unidad 29):** Divide en 2 partes **desiguales** usando un `pivot` (menores a la izquierda, mayores a la derecha), conquista ambas y **no requiere un proceso extra de combinación**.

Búsqueda Binaria: Iterativa vs. Recursiva

````python
# 1.1. Búsqueda Binaria Iterativa (Repaso Unidad 25)
def bin_search(nums, x):
    low, high = 0, len(nums) - 1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == x:
            return mid
        elif nums[mid] > x:
            high = mid - 1
        else:
            low = mid + 1
    return -1

# 1.2. Búsqueda Binaria Recursiva (Enfoque Divide y Conquista)
def bin_search2(nums, x, low, high):
    if low > high:
        return -1
    else:
        mid = (low + high) // 2
        if nums[mid] == x:
            return mid
        elif nums[mid] > x:
            return bin_search2(nums, x, low, mid - 1)
        else:
            return bin_search2(nums, x, mid + 1, high)

# Prueba con el ejemplo de las diapositivas: buscar x = 26
S = [11, 17, 26, 28, 37, 45, 53, 59, 77]
x = 26

print("Índice encontrado (Iterativo):", bin_search(S, x))
print("Índice encontrado (Recursivo D&C):", bin_search2(S, 53, 0, len(S) - 1))
````

El Problema de la Ruta Mínima del Triángulo

## 2. Problema de la Ruta Mínima del Triángulo (Triangle Path Finder)

Dado un triángulo de números enteros positivos, empezamos en la cima (`fila 0, columna 0`) y descendemos una fila a la vez.
En cada paso, solo podemos movernos hacia **abajo (`row + 1, col`)** o hacia **abajo a la derecha (`row + 1, col + 1`)**.
Queremos encontrar la ruta cuya suma de números sea el **costo mínimo**.

```text
       [2]          <- Fila 0
     [3,  4]        <- Fila 1
   [6,  5,  7]      <- Fila 2
 [4,  1,  8,  3]    <- Fila 3

Ruta óptima: 2 -> 3 -> 5 -> 1 = 11.

Ahora vamos a resolver la Misión de la Unidad 31: **La Ruta Mínima del Triángulo. Imaginen que este triángulo es un laberinto inclinado donde cada número es el peaje que deben pagar por pasar por esa habitación. Empezamos en el `2` arriba del todo. Desde cualquier número, solo tenemos dos movimientos permitidos: bajar recto (`row + 1, col`) o bajar en diagonal a la derecha (`row + 1, col + 1`).
  >¿Por qué no usamos **Greedy** aquí? Si fuéramos Greedy en un triángulo donde abajo del `2` hay un `3` y un `4`, elegiríamos el `3`. ¿Pero qué pasa si debajo del `3` hay puros números gigantes como `999` y debajo del `4` hay puros `1`? Greedy caería en la trampa.
  > Con **Divide y Conquista**, nos paramos en `(row, col)` y decimos: *'Mi costo mínimo es lo que cuesta mi casilla actual `T[row][col]` MÁS el mínimo entre resolver el sub-triángulo izquierdo y resolver el sub-triángulo derecho'.

````python
def find_minimum(row, col, triangle):
    if row == len(triangle):
        return 0
    else:
        minimum = min(
            find_minimum(row + 1, col, triangle),
            find_minimum(row + 1, col + 1, triangle)
        )
        return triangle[row][col] + minimum

triangle = [
    [2],
    [3, 4],
    [6, 5, 7],
    [4, 9, 8, 3]
]

minimum = find_minimum(0, 0, triangle)
print("The minimum cost is", minimum)
````

El Problema del Tromino

````python
# El Problema del Tromino (Tromino Puzzle)
# Un Tromino es una pieza en forma de "L" formada por 3 cuadrados.
# Dado un tablero de N x N (donde N es potencia de 2, ej. 4x4 u 8x8) con una casilla marcada con 'X' (-1),
# debemos cubrir todo el tablero con trominos sin superponerlos.

tromino_id = 0

def solve_tromino(board, size, top_row, top_col, missing_r, missing_c):
    global tromino_id
    # Caso base: si el sub-tablero es de 2x2, colocamos 1 tromino en las 3 casillas libres
    if size == 2:
        tromino_id += 1
        for r in range(top_row, top_row + 2):
            for c in range(top_col, top_col + 2):
                if r != missing_r or c != missing_c:
                    board[r][c] = tromino_id
        return

    # Paso 1: DIVIDIR el tablero actual en 4 cuadrantes de tamaño (size // 2)
    half = size // 2
    mid_r = top_row + half
    mid_c = top_col + half

    # Coordenadas de las 4 esquinas centrales donde se tocan los 4 cuadrantes
    center_cells = [
        (mid_r - 1, mid_c - 1),  # Cuadrante Superior-Izquierdo
        (mid_r - 1, mid_c),      # Cuadrante Superior-Derecho
        (mid_r, mid_c - 1),      # Cuadrante Inferior-Izquierdo
        (mid_r, mid_c)           # Cuadrante Inferior-Derecho
    ]

    # Identificar en cuál de los 4 cuadrantes está la casilla ocupada/'X' actual
    quad_with_missing = (0 if missing_r < mid_r else 2) + (0 if missing_c < mid_c else 1)

    # Colocar 1 Tromino central que ocupe una esquina de los 3 cuadrantes que NO tienen 'X'
    tromino_id += 1
    current_id = tromino_id
    for q in range(4):
        if q != quad_with_missing:
            cr, cc = center_cells[q]
            board[cr][cc] = current_id

    # Paso 2: CONQUISTAR recursivamente cada uno de los 4 sub-cuadrantes
    for q in range(4):
        q_top_r = top_row if q < 2 else mid_r
        q_top_c = top_col if q % 2 == 0 else mid_c
        if q == quad_with_missing:
            solve_tromino(board, half, q_top_r, q_top_c, missing_r, missing_c)
        else:
            cr, cc = center_cells[q]
            solve_tromino(board, half, q_top_r, q_top_c, cr, cc)

def print_board(board):
    for row in board:
        print(" ".join(f"{ 'X' if val == -1 else str(val):>2}" for val in row))

# Prueba en el tablero 4x4 de las diapositivas (X en fila 0, columna 1)
N = 4
board = [[0] * N for _ in range(N)]
x_row, x_col = 0, 1
board[x_row][x_col] = -1  # Marcamos la X con -1

solve_tromino(board, N, 0, 0, x_row, x_col)
print(f"Tablero {N}x{N} resuelto (cada número es un Tromino en forma de L):")
print_board(board)
````
